```python
import numpy as np

T = np.array([97.674, 104.070, 111.628, 113.372, 125.000, 136.047, 131.977])

n = np.array([1.237, 1.235, 1.229, 1.228, 1.222, 1.215, 1.216])
```

```python
import matplotlib.pyplot as plt

coeffs = np.polyfit(T, n, 1)
fit    = np.poly1d(coeffs)

T_ext  = 148.0
n_ext  = fit(T_ext)
print(f"Linear fit: n = {coeffs[0]:.6f} · T + {coeffs[1]:.6f}")
print(f"n at T = {T_ext} K  →  {n_ext:.4f}")

T_line = np.linspace(T.min() - 5, T_ext + 5, 200)
fig, ax = plt.subplots()
ax.scatter(T, n, zorder=3, label='data')
ax.plot(T_line, fit(T_line), label='linear fit')
ax.scatter(T_ext, n_ext, marker='*', s=120, zorder=4, label=f'T={T_ext} K → n={n_ext:.4f}')
ax.set_xlabel('T (K)')
ax.set_ylabel('n')
ax.legend()
plt.tight_layout()
plt.show()
```
